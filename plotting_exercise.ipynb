### Execise 4

Write a loop that iterates over these example datasets and plots a simple scatter plot of each dataset with the `x` variable on the x-axis and the `y` variable on the y-axis.

````python
for i in range(1, 14):
    url = f"https://raw.githubusercontent.com/nickeubank/practicaldatascience_book/refs/heads/main/notebooks/class_5/week_1/data/basic_plotting_datasets/datasaurus_{i}.txt"
    df = pd.read_csv(url, delimiter="\t")

    plt.figure(figsize=(6, 4))
    plt.scatter(df["x_var"], df["y_var"])
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(f"Example Dataset {i}")
    plt.show()
````

### Exercise 6

Load the World Development Indicator data [here](https://raw.githubusercontent.com/nickeubank/practicaldatascience/master/Example_Data/wdi_plotting.csv)

Rather than picking a single year, pick a single country and look at how GDP per capita and one of the other variables in that dataset have evolved together over time. For now, plot the two series them as two separate plots. Be sure your figure includes a title and x and y-axis labels in plain English (not variable names).

Make any adjustments to the functional forms of your variables and/or axes needed to make the figure legible. 

````python
url_wd = "https://raw.githubusercontent.com/nickeubank/practicaldatascience/master/Example_Data/wdi_plotting.csv"

df_wd = pd.read_csv(url_wd)

print(df_wd.sample(2))
````

````python
# subset dr_wd by "Country Name" == "China"
chn_development = df_wd[df_wd["Country Name"] == "China"].sort_values("Year")

fig, axes = plt.subplots(2, 1, figsize=(9, 8), sharex=True)

axes[0].plot(
    chn_development["Year"],
    # divide GDP by 1000 for ease of axis label readability
    chn_development["GDP per capita (constant 2010 US$)"] / 1000,
)
axes[0].set_title("GDP per Capita of China Over Time")
axes[0].set_xlabel("Year")
axes[0].set_ylabel("GDP per person\n(thousands of 2010 US dollars)")

axes[1].plot(
    chn_development["Year"], chn_development["Population, total"], color="orange"
)
axes[1].set_title("Population of China Over Time")
axes[1].set_xlabel("Year")
axes[1].set_ylabel("Population (billions)")

for ax in axes:
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

### Exercise 7

Let's put the two series together so that they share a common x-axis of (so both series are plotted in the same plot, and the y-axes is time. Obviously your different series will need different y-axes). 

Use your detective skills (and some guess and check work) to figure out how to get it to work!


````python
fig, ax1 = plt.subplots(figsize=(9, 5))
ax2 = ax1.twinx()

(line1,) = ax1.plot(
    chn_development["Year"],
    chn_development["GDP per capita (constant 2010 US$)"] / 1000,
    color="blue",
    label="GDP per capita",
)

(line2,) = ax2.plot(
    chn_development["Year"],
    chn_development["Population, total"] / 1_000_000_000,
    color="red",
    label="Population",
)

ax1.set_xlabel("Year")
ax1.set_ylabel("GDP per person (thousands of 2010 US dollars)", color="blue")
ax2.set_ylabel("Population (billions)", color="red")

ax1.tick_params(axis="y", labelcolor="blue")
ax2.tick_params(axis="y", labelcolor="red")

ax1.set_title("China: GDP per Capita and Population Over Time")
ax1.legend(handles=[line1, line2], loc="upper left")
ax1.grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

### Exercise 8

LLMs are *great* for plotting, and they're really good at matplotlib (it's been around forever, so there's lots of training data). Using Github Copilot Chat in VS Code, made the following changes to your plot:

- Add a note to the bottom of the plot in the bottom left that reports the source for your data.
- Change the x-axes so that it prints out the year every 5 years (1990, 1995, 2000, etc.).
- Add a red, dashed, vertical line at the year 2000.
- Add text that is rotated 90 degrees, in red, alongside that vertical line in the middle of the plot that says "New Millenium."

````python
from matplotlib.ticker import MultipleLocator, StrMethodFormatter

fig, ax1 = plt.subplots(figsize=(9, 5))
ax2 = ax1.twinx()

(line1,) = ax1.plot(
    chn_development["Year"],
    chn_development["GDP per capita (constant 2010 US$)"] / 1000,
    color="blue",
    label="GDP per capita",
)

(line2,) = ax2.plot(
    chn_development["Year"],
    chn_development["Population, total"] / 1_000_000_000,
    color="brown",
    label="Population",
)

ax1.set_xlabel("Year")
ax1.xaxis.set_major_locator(MultipleLocator(5))
ax1.xaxis.set_major_formatter(StrMethodFormatter("{x:.0f}"))
ax1.set_ylabel("GDP per person (thousands of 2010 US dollars)", color="blue")
ax2.set_ylabel("Population (billions)", color="brown")

ax1.tick_params(axis="y", labelcolor="blue")
ax2.tick_params(axis="y", labelcolor="brown")

ax1.set_title("China: GDP per Capita and Population Over Time")
ax1.legend(handles=[line1, line2], loc="upper left")
ax1.grid(alpha=0.3)

ax1.axvline(2000, color="red", linestyle="--")
ax1.annotate(
    "New Millenium",
    xy=(2000, 0.5),
    xycoords=ax1.get_xaxis_transform(),
    xytext=(0, 0),
    textcoords="offset points",
    rotation=90,
    color="red",
    ha="right",
    va="center",
)

fig.text(
    0.02,
    0.02,
    "Source: World Bank, World Development Indicators (wdi_plotting.csv).",
    ha="left",
    va="bottom",
    fontsize=9,
)
plt.tight_layout(rect=(0, 0.07, 1, 1))
plt.show()
````
